## 1. Introdução
### Objetivo do trabalho

Desenvolver um Sistema de Processamento Visual (SPV) capaz de monitorar, em tempo real, a postura de perfil de usuários sentados, utilizando visão computacional para detectar desalinhamentos posturais e fornecer alertas imediatos.

### Motivação

Desvios posturais prolongados podem causar:

* dores musculares
* problemas ergonômicos
* lesões na coluna
* redução da qualidade de vida

O sistema proposto visa auxiliar usuários comuns a corrigirem hábitos inadequados de postura.

## 2. Cenário de Aplicação (CA)

* Correção em tempo real

## 3. Fundamentação Teórica
A fundamentação teórica deste projeto está baseada na aplicação integrada de conceitos clássicos e modernos de Processamento Digital de Imagens e Visão Computacional, com foco na análise postural em tempo real. Inicialmente, utiliza-se o processamento de imagens por meio da biblioteca OpenCV, responsável pela captura, leitura, conversão e manipulação das imagens e vídeos analisados pelo sistema. O processamento de cores desempenha papel essencial, especialmente na conversão entre os espaços BGR e RGB, necessária para compatibilidade entre OpenCV e MediaPipe. Além disso, técnicas de análise de histograma e contraste são empregadas para avaliar a distribuição de intensidades das imagens antes e após o processamento, permitindo compreender alterações visuais introduzidas pela marcação dos pontos corporais e pela segmentação visual. O sistema também incorpora conceitos de segmentação corporal e extração de características, utilizando a biblioteca MediaPipe Pose Landmarker para identificar landmarks anatômicos relevantes, como nariz, ombros e quadris, que servem como base para a modelagem geométrica da postura do usuário. A partir desses pontos, aplica-se análise espacial e cálculo angular para estimar o alinhamento corporal, permitindo classificar a postura como correta ou inadequada. Complementarmente, operadores de processamento visual, como filtragem e melhorias de contraste, contribuem para maior robustez na detecção, especialmente em condições variadas de iluminação e posicionamento. Dessa forma, o projeto combina fundamentos matemáticos, geométricos e computacionais para criar uma solução prática, interativa e acessível de monitoramento postural, atendendo aos requisitos técnicos da disciplina e demonstrando a aplicação real dos conteúdos estudados.

## 8. Conclusões
O desenvolvimento do Sistema de Processamento Visual (SPV) para monitoramento postural permitiu alcançar de forma satisfatória os objetivos propostos pelo trabalho, demonstrando a aplicação prática dos conceitos estudados na disciplina de Processamento Digital de Imagens. A solução implementada mostrou-se funcional ao utilizar técnicas de visão computacional, processamento de imagens e análise geométrica corporal para monitorar, em tempo real, a postura de usuários sentados, fornecendo feedback imediato sobre possíveis desvios posturais. O sistema apresentou boa usabilidade, sendo facilmente operado por usuários leigos, conforme validado durante os testes de campo e pelas avaliações subjetivas obtidas. Entre os principais pontos positivos destacam-se a simplicidade de execução, a interface intuitiva, a capacidade de processamento em tempo real e a relevância prática da aplicação no contexto de saúde preventiva e ergonomia. Além disso, o projeto evidenciou criatividade ao propor uma ferramenta de utilidade cotidiana, voltada à prevenção de problemas físicos decorrentes de hábitos posturais inadequados. Contudo, algumas limitações foram observadas, como a dependência de condições adequadas de iluminação, o correto posicionamento da câmera e possíveis variações na precisão da detecção em diferentes cenários de uso. Como perspectivas futuras, o sistema pode ser aprimorado com técnicas mais avançadas de calibração angular, maior robustez em ambientes diversos, expansão para múltiplos usuários e integração com mecanismos automáticos de alerta mais sofisticados. De modo geral, conclui-se que o projeto atendeu aos requisitos técnicos e experimentais estabelecidos, consolidando-se como uma aplicação eficiente, funcional e academicamente relevante dentro da proposta da disciplina.
